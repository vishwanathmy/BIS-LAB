```python

import random
import math

cities = {
    'A': (0, 0),
    'B': (2, 3),
    'C': (5, 2),
    'D': (6, 6),
    'E': (1, 5)
}

N = 8
GEN = 20
MR = 0.2
CR = 0.8

names = list(cities.keys())

def distance(a, b):
    x1, y1 = cities[a]
    x2, y2 = cities[b]
    return math.sqrt((x2-x1)**2 + (y2-y1)**2)

def expression(gene):
    return gene + [gene[0]]

def fitness(gene):
    r = expression(gene)
    d = sum(distance(r[i], r[i+1]) for i in range(len(r)-1))
    return 1/d

def crossover(p1, p2):
    a, b = sorted(random.sample(range(len(names)), 2))
    child = p1[a:b] + [x for x in p2 if x not in p1[a:b]]
    return child

def mutation(child):
    if random.random() < MR:
        a, b = random.sample(range(len(names)), 2)
        child[a], child[b] = child[b], child[a]
    return child

pop = [random.sample(names, len(names)) for _ in range(N)]

for g in range(GEN):
    pop.sort(key=fitness, reverse=True)
    new = [pop[0]]

    while len(new) < N:
        p1, p2 = random.choices(pop[:4], k=2)

        if random.random() < CR:
            child = crossover(p1, p2)
        else:
            child = p1.copy()

        new.append(mutation(child))

    pop = new

best = max(pop, key=fitness)

print("Best Route:", " -> ".join(expression(best)))
print("Minimum Distance:", round(1/fitness(best), 2))
```
